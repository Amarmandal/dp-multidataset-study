```python
import json

with open("./data/dp_lr_report.json", "r") as f:
    data = json.load(f)
```

```python
epsilon_results = data["epsilon_results"]

dp_rf = {
    "epsilon_values": [],
    "accuracy": [],
    "precision": [],
    "recall": [],
    "f1_score": []
}

for eps, result in epsilon_results.items():
    dp_rf["epsilon_values"].append(float(eps))
    dp_rf["accuracy"].append(result["accuracy"])
    
    malignant_metrics = result["classification_report"]["Malignant"]
    
    dp_rf["precision"].append(malignant_metrics["precision"])
    dp_rf["recall"].append(malignant_metrics["recall"])
    dp_rf["f1_score"].append(malignant_metrics["f1-score"])

# Optional: sort by epsilon (recommended for plotting)
sorted_indices = sorted(range(len(dp_rf["epsilon_values"])),
                        key=lambda i: dp_rf["epsilon_values"][i])

dp_rf = {
    key: [values[i] for i in sorted_indices]
    for key, values in dp_rf.items()
}

print(dp_rf)
```
